```python
import chromadb
chroma_client = chromadb.Client()
```

```python
collection = chroma_client.create_collection(name="my_collection")
```

```python
collection.add(
    documents=[
        "This is a document about Delhi",
        "This is a document about Assam"
    ],
    ids=["id3", "id4"]
)
```

```python
all_docs=collection.get()
all_docs
```

```python
results = collection.query(
    query_texts=["Query is about Capital of India"], # Chroma will embed this for you
    n_results=2
)
print(results)
```

```python
#Deleting Documents
collection.delete(ids=all_docs['ids'])
collection.get()
```
